## 📚 Importing Required Libraries

Load the Python libraries required for data manipulation, visualization, preprocessing, model training, and evaluation.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,root_mean_squared_error
```

Read the Bangalore housing dataset into a Pandas DataFrame for analysis.

```python
df = pd.read_csv('W:/Data-Science-Main/Data/bangalore_v2.csv')
```

## 👀 Dataset Preview

Inspect the dataset structure and verify that the data has been loaded correctly.

```python
df = pd.get_dummies(df, columns=['location'], drop_first=True)
```

## 👀 Dataset Preview

Inspect the dataset structure and verify that the data has been loaded correctly.

```python
x = df.drop('price_lakh', axis=1)
y = df['price_lakh']
```

## ✂️ Train-Test Split

Split the dataset into training and testing subsets for unbiased evaluation.

```python
xTrain, xTest, yTrain, yTest = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

## 🤖 Model Initialization

Create a Linear Regression model instance.

```python
model = LinearRegression()
```

## 🚀 Model Training

Train the Linear Regression model using the training dataset.

```python
model.fit(xTrain,yTrain)
```

## 🔮 Generating Predictions

Use the trained model to predict housing prices.

```python
predict = model.predict(xTest)
```

## 🔮 Generating Predictions

Use the trained model to predict housing prices.

```python
mse = mean_squared_error(yTest, predict)

print("MSE :", mse)

rmse = root_mean_squared_error(yTest, predict)

print("RMSE :", rmse)

r2 = (r2_score(yTest, predict) * 100).__round__(2)

print("R² :", r2,"%")
```

## 📊 Data Visualization

Visualize patterns, distributions, or model performance.

```python
plt.style.use('dark_background')

plt.figure(figsize=(10,7))

plt.scatter(
    yTest,
    predict,
    alpha=1,
    color='black',
    edgecolors='green'
)

min_val = min(yTest.min(), predict.min())
max_val = max(yTest.max(), predict.max())

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    color='red',
    linewidth=2,
    label='Ideal Fit'
)

plt.xlabel("Actual Price (Lakhs)")
plt.ylabel("Predicted Price (Lakhs)")
plt.title("Actual vs Predicted Housing Prices")
plt.legend()
plt.grid(alpha=0.3, linestyle=':')

plt.show()
```

## 🔮 Generating Predictions

Use the trained model to predict housing prices.

```python
residuals = yTest - predict
```

## 📊 Data Visualization

Visualize patterns, distributions, or model performance.

```python
plt.figure(figsize=(8,6))

plt.scatter(predict, residuals, alpha=1,color='black',
    edgecolors='green',)

plt.axhline(y=0, color='red')

plt.xlabel("Predicted Price")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.grid(alpha=0.3, linestyle=':')

plt.show()
```
